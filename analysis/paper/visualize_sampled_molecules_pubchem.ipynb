# Visualize Sampled Molecules

This notebook visualizes the molecules sampled from PubChem.

```python
from rdkit import Chem
from rdkit.Chem import Draw
from rdkit.Chem.rdCoordGen import AddCoords
from IPython.display import display

# Read SMILES from file
txt_path = "../../data/molecules/sampled_molecules.txt"

with open(txt_path, "r") as f:
    smiles_list = [line.strip() for line in f if line.strip()]

print(f"Loaded {len(smiles_list)} molecules")
```

```python
# Convert SMILES to molecules
mols = [Chem.MolFromSmiles(smi) for smi in smiles_list]
mols = [mol for mol in mols if mol is not None]  # Filter out None

print(f"Successfully parsed {len(mols)} molecules")
```

```python
for mol in mols:
    AddCoords(mol)
```

```python
import os
from rdkit.Chem import Draw
import cairosvg  

output_dir = "figures"
os.makedirs(output_dir, exist_ok=True)

pdf_path = os.path.join(output_dir, "sampled_molecules_grid.pdf")

svg = Draw.MolsToGridImage(
    mols,
    molsPerRow=10,
    subImgSize=(200, 200),
    legends=[f"Mol {i+1}" for i in range(len(mols))],
    useSVG=True
).data  # SVG string

cairosvg.svg2pdf(bytestring=svg.encode("utf-8"), write_to=pdf_path)
print(f"Saved vector PDF to {pdf_path}")
```
