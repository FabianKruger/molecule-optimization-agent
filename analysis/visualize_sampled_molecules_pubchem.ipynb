# Visualize Sampled Molecules

This notebook visualizes the molecules sampled from PubChem.

```python
from rdkit import Chem
from rdkit.Chem import Draw
from IPython.display import display

# Read SMILES from file
txt_path = "sampled_molecules.txt"

with open(txt_path, "r") as f:
    smiles_list = [line.strip() for line in f if line.strip()]

print(f"Loaded {len(smiles_list)} molecules")
```

```python
# Convert SMILES to molecules
mols = [Chem.MolFromSmiles(smi) for smi in smiles_list]
mols = [mol for mol in mols if mol is not None]  # Filter out None

print(f"Successfully parsed {len(mols)} molecules")
```

```python
# Visualize molecules in a grid
img = Draw.MolsToGridImage(
    mols,
    molsPerRow=4,
    subImgSize=(300, 300),
    legends=[f"Mol {i+1}" for i in range(len(mols))]
)
display(img)
```
